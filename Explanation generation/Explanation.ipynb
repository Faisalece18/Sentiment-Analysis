```python
import google.generativeai as genai
import pandas as pd
import json
import time
```

```python

genai.configure(api_key="")
```

```python

model = genai.GenerativeModel("gemini-2.0-flash")
```

```python
df = pd.read_excel('combined_data_20.xlsx')
```

```python
system_prompt = """
You are a highly knowledgeable Bengali linguistic expert and sentiment analyst.
Your goal is to carefully analyze the given Bangla text and generate a detailed explanation for the provided sentiment label.

Instructions:
- You will be given a Bangla text and its sentiment (Positive, Negative, or Neutral).
- DO NOT predict or change the sentiment. Focus only on explaining why the given sentiment is appropriate.
- You must provide a concrete, clear, and well-supported explanation why the text belongs to that sentiment.

When generating explanations, always consider:

✅ Textual Content  
- Identify emotional words, expressions, or structures that support the sentiment.  
- Highlight important words, phrases, or punctuation that contribute to the emotional tone.

🧠 Linguistic Features  
- Analyze the use of adjectives, adverbs, verb tenses, negations ("না", "নয়", "কখনো না"), and contrast markers ("কিন্তু", "তবে").  
- Discuss sentence structure, complexity, and compound words if relevant.

📚 Semantic Understanding  
- Understand the deeper contextual meaning.  
- Detect subtle forms of sarcasm, irony, hopefulness, criticism, or neutrality.

🎭 Tone and Overall Context  
- Comment on whether the tone is formal, humorous, sarcastic, optimistic, concerned, or neutral.  
- Explain how the tone and context align with the provided sentiment.

Special Focus:
- Always consider tone, context, semantic meaning, and emotionally loaded words—especially in political, social, or economic discussions.

📝 Output Format: 
Explanation (in Bangla): [expert reasoning in Bengali]
"""
```

```python
def generate_explanation(text, sentiment, model):
    prompt = f"""
{system_prompt}

Input Text (Bangla): {text}
Given Sentiment: {sentiment}
"""
    try:
        response = model.generate_content(prompt)
        explanation = response.text.strip()
        return explanation
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
results = []
```

```python
for idx, row in df.iterrows():
    text = row["text"]
    sentiment = row["Tag"].capitalize()  

    print(f"\n🔍 Generating explanation {idx+1}/{len(df)}:")

    explanation = generate_explanation(text, sentiment, model)

    # 📢 Print the explanation immediately
    print(f"📜 Explanation:\n{explanation}")

    
    results.append({
        "sample": idx+1,
        "Input Text (Bangla):": text,
        "Given Sentiment": sentiment,
        "explanation": explanation
    })

    time.sleep(1.5) 
```

```python
with open("bangla_sentiment_explanations.json", "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=4)

print("\n✅ All explanations saved to 'bangla_sentiment_explanations.json'")
```
